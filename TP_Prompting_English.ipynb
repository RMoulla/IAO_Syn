# Hands-on Lab — Prompting

## Objectives
- Design clear prompts with measurable constraints.
- Obtain outputs in a strict format (sections, tables, JSON).
- Adapt tone and structure to a specific situation.
- Iterate and make tangible improvements to an output.


## A universal prompt structure
1. **Context & audience** – who the output is for and in what situation.
2. **Role** – the role the AI should adopt.
3. **Task** – a single, observable action.
4. **Constraints** – length, tone, prohibited elements, data to include.
5. **Output format** – sections, tables, JSON, etc.
6. **Verification** – flag uncertainties / decline if the requirements cannot be met.


## **Exercise 1 — LinkedIn post**
**Objective:** obtain text that follows a precise template.

Prompt to copy and paste

```prompt
You are a communications manager. Write a professional LinkedIn post announcing the upcoming Food Innovation Trade Show,
where your company will present its new data analytics solutions for the retail sector.
Length constraint: 200–300 words.
Tone: clear and engaging, without superlatives.
Exact structure:
1) Hook (1 sentence, ≤ 20 words)
2) Context (4 sentences)
3) Value for visitors (3 bullet points)
4) Practical details (date, venue, booth, opening hours)
5) Call to action (a simple, direct CTA)
Prohibited: excessive technical jargon, vague promises, excessive emojis.
Output format: strictly follow the 5 sections in the specified order.
```

Prompt to copy and paste (after the response)

```prompt
Critically assess the text you generated. Identify three strengths and three weaknesses.
```


## **Exercise 3 — Extracting information from unstructured text**
**Objective:** produce usable JSON from noisy text without inventing information.

```context
“So, these *Borea X-2* headphones (yes, some people write X2) are our current favorite wireless headphones. Battery life is close to 30 h (in the lab: 28 h 30 min). Bluetooth 5.3, built-in ANC, USB-C, weight around 240 g. Store price seen yesterday: €129 including VAT (there was a €119 promotion, but it has ended). Advertised warranty: 2 years. Really great… except for the mediocre microphone and the very basic mobile app. Colors: black/blue. Delivery within 48 h if in stock.”
```

Prompt to copy and paste

```prompt
Role: meticulous data specialist.
Task: extract the product information from the text above as strict JSON on a single line.
Schema:
`{"product":"","price_eur":0,"battery_life_h":0,"weight_g":0,"connectivity":"","anc":false,"warranty_years":0,"strengths":[],"weaknesses":[]}`
Rules:
- Do not invent anything; if information is missing, use `null`.
- Conflicts: prioritize the nominal specification (e.g. “30 h”) over the lab measurement (“28 h 30 min”).
- Normalization:
  - `connectivity` = “Bluetooth 5.3”.
  - `anc` = `true` if ANC is mentioned; otherwise, `false`.
  - `price_eur` = the regular retail price excluding promotions (129).
  - `weight_g` = an integer (240).
  - `battery_life_h` = a number (30).
Output format: return only the JSON line, nothing else.
```
